```python
import sys

input = sys.stdin.readline

def main():
    N, Q = map(int, input().split())
    data = list(map(int, input().split()))
    prefix_sum = [0] * (N + 1)
    prefix_sq_sum = [0] * (N + 1)

    for i in range(N):
        prefix_sum[i+1] = prefix_sum[i] + data[i]
        prefix_sq_sum[i+1] = prefix_sq_sum[i] + data[i] * data[i]

    for _ in range(Q):
        parts = input().split()
        if parts[0] == '1':
            k = int(parts[1])
            n = len(data)
            sum_k = prefix_sum[n] - prefix_sum[n - k]
            sq_sum_k = prefix_sq_sum[n] - prefix_sq_sum[n - k]
            mean = sum_k / k
            variance = (sq_sum_k - k * mean * mean) / k
            print(f"{variance:.7f}")
        elif parts[0] == '2':
            x = int(parts[1])
            data.append(x)
            new_index = len(data)
            prefix_sum.append(prefix_sum[new_index - 1] + x)
            prefix_sq_sum.append(prefix_sq_sum[new_index - 1] + x * x)

main()
```
