```python
from scipy.stats import beta

# Параметры Beta(3, 3)
a, b = 3, 3

# Вероятность, что x_3 попадает в [15, 25]
p = beta.cdf(0.625, a, b) - beta.cdf(0.375, a, b)
print(p)  # ≈ 0.682
```

```python
import numpy as np

def simulate_probability(num_simulations=1_000_000):
    count_success = 0
    
    for _ in range(num_simulations):
        # Генерируем 5 случайных точек и сортируем их
        split_points = np.sort(np.random.uniform(0, 40, 5))
        
        # Третья точка разбиения (U_{(3)})
        u3 = split_points[2]
        
        # Проверяем условие 15 ≤ U₃ ≤ 25
        if 15 <= u3 <= 25:
            count_success += 1
    
    probability = count_success / num_simulations
    return probability

# Запускаем симуляцию
prob = simulate_probability()
print(f"Вероятность: {prob:.4f} (или ~{prob*100:.2f}%)")
```

```python
import numpy as np

# Параметры
n_simulations = 1000000  # Количество симуляций
T = 40  # Длительность альбома
max_side = 25  # Максимальная длительность стороны
n_splits = 5  # Количество разделителей

# Генерация случайных разделителей
np.random.seed(42)  # Для воспроизводимости
split_points = np.random.uniform(0, T, (n_simulations, n_splits))

# Сортировка точек для получения порядковых статистик
split_points.sort(axis=1)

# Проверка условия: третья точка <= 25
success = split_points[:, 2] <= max_side  # T_(3) <= 25
probability = np.mean(success)

print(f"Оценочная вероятность: {probability:.6f}")
```

```python
import numpy as np

def estimate_probability(num_trials=1_000_000):
    count_success = 0

    for _ in range(num_trials):
        cuts = np.sort(np.random.uniform(0, 40, size=5))
        intervals = np.diff(np.concatenate(([0], cuts, [40])))
        if intervals[:3].sum() <= 25:
            count_success += 1

    return count_success / num_trials

prob = estimate_probability()
print(f"Вероятность: {prob:.5f}")
```

```python
import numpy as np

# Параметры
total_duration = 40  # длительность альбома в минутах
max_side_duration = 25  # максимальная длительность одной стороны
num_splits = 5  # количество разделителей
num_songs_first_side = 3  # количество песен на первой стороне
num_simulations = 1000000  # количество симуляций

# Моделирование
np.random.seed(42)  # для воспроизводимости
success_count = 0

for _ in range(num_simulations):
    # Генерируем 5 случайных разделителей
    splits = np.random.uniform(0, total_duration, num_splits)
    # Сортируем разделители
    sorted_splits = np.sort(splits)
    # Проверяем, что время до третьего разделителя <= 25 минут
    if sorted_splits[num_songs_first_side - 1] <= max_side_duration:
        success_count += 1

# Вероятность
probability = success_count / num_simulations
print(f"Вероятность: {probability:.4f}")
```

```python
import numpy as np

# Параметры симуляции
n_trials = 1_000_000
album_length = 40
max_side_length = 25

# Генерация 5 случайных точек для каждой симуляции
points = np.sort(np.random.uniform(0, album_length, size=(n_trials, 5)), axis=1)

# Добавляем начала и конца альбома
points = np.concatenate([
    np.zeros((n_trials, 1)),  # 0
    points,
    np.full((n_trials, 1), album_length)  # 40
], axis=1)

# Вычисляем длины треков
track_lengths = np.diff(points, axis=1)

# Считаем сумму первых трёх и последних трёх
side1 = track_lengths[:, :3].sum(axis=1)
side2 = track_lengths[:, 3:].sum(axis=1)

# Условие: обе стороны ≤ 25 минут
successes = np.logical_and(side1 <= max_side_length, side2 <= max_side_length)

# Вероятность успеха
probability = np.mean(successes)
probability
```
