```python
def count_symmetric_splits(S):
    if len(S) < 2:
        return -1

    left_mask = 0
    right_mask = 0
    result_count = 0

    for ch in S:
        digit = int(ch)
        right_mask |= (1 << digit)

    for i in range(len(S)):
        digit = int(ch)
        digit = int(S[i])

        left_mask |= (1 << digit)
        if S.find(S[i], i + 1) == -1:
            right_mask &= ~(1 << digit)

        if left_mask == right_mask:
            result_count += 1

    return result_count if result_count > 0 else -1

S = input().strip()
print(count_symmetric_splits(S))
```
